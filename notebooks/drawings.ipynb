```python
import FreeCAD
import FreeCADGui
import Mesh
import Part
import TechDraw

def generate_technical_drawing(stl_file, output_file):
    # Create a new FreeCAD document
    doc = FreeCAD.newDocument("Drawing")

    # Import the STL file as a mesh
    mesh_obj = Mesh.Mesh(stl_file)
    mesh_feature = doc.addObject("Mesh::Feature", "Mesh")
    mesh_feature.Mesh = mesh_obj

    # Convert the mesh into a shape
    shape = Part.Shape()
    shape.makeShapeFromMesh(mesh_obj.Topology, 0.1)
    solid = Part.Solid(shape)

    # Create a new Part object
    obj = doc.addObject("Part::Feature", "ImportedPart")
    obj.Shape = solid

    # Create a TechDraw page
    page = doc.addObject("TechDraw::DrawPage", "Page")
    template = doc.addObject("TechDraw::DrawSVGTemplate", "Template")

    # Use the correct TechDraw template (A4_Landscape_TD)
    template.Template = FreeCAD.getResourceDir() + "Mod/TechDraw/Templates/A4_Landscape_TD.svg"
    page.Template = template

    # Add standard views
    for direction, name in [((0, -1, 0), "Front"), ((0, 0, 1), "Top"), ((1, 0, 0), "Side")]:
        view = doc.addObject("TechDraw::DrawViewPart", f"{name}View")
        view.Source = obj
        view.Direction = direction
        page.addView(view)

    # Add dimensions (horizontal and vertical) - example
    dim_h = doc.addObject("TechDraw::DrawDimHorizontal", "DimH")
    dim_h.References2D = [(page.Views[0], "Vertex1"), (page.Views[0], "Vertex2")]  # Adjust for specific edges/vertices
    page.addView(dim_h)

    dim_v = doc.addObject("TechDraw::DrawDimVertical", "DimV")
    dim_v.References2D = [(page.Views[1], "Vertex1"), (page.Views[1], "Vertex2")]  # Adjust for specific edges/vertices
    page.addView(dim_v)

    # Recompute document
    doc.recompute()
    doc.saveAs(output_file)
    print(f"✅ Technical drawing saved as: {output_file}")

# Example usage
generate_technical_drawing("yagi-uda-mount.stl", "output.FCStd")
```
